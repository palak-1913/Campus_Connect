**Let's Convert Text To vector**

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create TF-IDF vectorizer
tfidf = TfidfVectorizer( stop_words='english')
```

```python
# Converting for data_2
tfidf_matrix = tfidf.fit_transform(data_2['Combined'])

feature_names = tfidf.get_feature_names_out()

print(f"TF-IDF Matrix shape: {tfidf_matrix.shape}")
print(f"Total features (unique skills/projects): {len(feature_names)}")
print(f"\nSample features: {feature_names[:20]}")
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

for k in range(2, 14):
    model = KMeans(n_clusters=k, random_state=42)
    labels = model.fit_predict(tfidf_matrix)
    score = silhouette_score(tfidf_matrix, labels)
    print(f"K={k}, Score={score}")
```

```python

inertia = []
k_values = range(2, 13)

for k in k_values:
    model = KMeans(n_clusters=k, random_state=42)
    model.fit(tfidf_matrix)
    inertia.append(model.inertia_)

plt.figure(figsize=(8,5))
plt.plot(k_values, inertia, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.xticks(k_values)
plt.grid()

plt.show()
```

```python
kmeans = KMeans(n_clusters=6, random_state=42)
clusters_2 = kmeans.fit_predict(tfidf_matrix)

data_2['cluster'] = clusters_2

print("Cluster Distribution:")
print(data_2['cluster'].value_counts().sort_index())

# Print cluster centers
print(f"\nCluster centers shape: {kmeans.cluster_centers_.shape}")
print(f"Inertia (within-cluster sum of squares): {kmeans.inertia_:.2f}")
```

```python
data['cluster'] = data_2['cluster']
data.head()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def get_recommendations_within_cluster(id, data_df, tfidf_matrix, kmeans_model, top_n=10):
                                      
    student_cluster = data_df.loc[id, 'cluster']
    cluster_students = data_df[data_df['cluster'] == student_cluster].index.tolist()

    student_vector = tfidf_matrix[id]

    cluster_vectors = tfidf_matrix[cluster_students]
    cosine_sim = cosine_similarity(student_vector, cluster_vectors).flatten()


    projects_normalized = data_df.loc[cluster_students, 'Number of Projects'].values / 10.0

    # Calculating score
    scores = (0.9 * cosine_sim) + (0.1 * projects_normalized)

    # Create results dataframe
    results = pd.DataFrame({
        'id': cluster_students,
        'score': scores,
        'cosine_sim': cosine_sim,
    })

    # getting top N
    results = results.sort_values('score', ascending=False).head(top_n)


    return results
```

```python
def get_recomendation_in_your_cluster(id,n):
    final_recomendation = get_recommendations_within_cluster(id=id, data_df=data, tfidf_matrix=tfidf_matrix, top_n=n+1,kmeans_model=kmeans)
    
    final_recomendation = final_recomendation.sort_values('score',ascending = False).head(n).reset_index(drop=True)
    ids = final_recomendation['id'].tolist()
    top_students = data[data['id'].isin(ids)]
    return top_students 
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def get_recommendations_among_all(id, df, tfidf_matrix, kmeans_model = kmeans, top_n=10):

    student_vector = tfidf_matrix[id]

    cosine_sim = cosine_similarity(student_vector, tfidf_matrix).flatten()


    projects_normalized = df['Number of Projects'].values / 10.0

    # Calculate composite score
    scores = (0.9 * cosine_sim) + (0.1 * projects_normalized)

    results = pd.DataFrame({
        'id': df.index,
        'score': scores,
        'cosine_sim': cosine_sim,

    })


    # Sort by score (descending) and get top N
    results = results.sort_values('score', ascending=False).head(top_n)


    return results
```

```python
def get_recomendation_overall(id,n):
    final_recomendation = get_recommendations_among_all(id=id, df=data, tfidf_matrix=tfidf_matrix, top_n=n+1,kmeans_model=kmeans)
    
    final_recomendation = final_recomendation.sort_values('score',ascending = False).head(n).reset_index(drop=True)
    ids = final_recomendation['id'].tolist()
    top_students = data[data['id'].isin(ids)]
    return top_students 
```

```python
data_sent = raw_data.copy()
data_sent.head()
```

```python
from sentence_transformers import SentenceTransformer

sent_transformer = SentenceTransformer('all-MiniLM-L6-v2')

data_sent['text'] = data_sent['Skills'] + " " + data_sent['Projects']

embeddings = sent_transformer.encode(data_sent['text'].tolist(), show_progress_bar=True)
```

```python
def normalize_features(df):
    df['proj_norm'] = (df['Number of Projects'] - df['Number of Projects'].min()) / (
        df['Number of Projects'].max() - df['Number of Projects'].min() + 1e-8
    )

    df['cgpa_norm'] = (df['CGPA'] - df['CGPA'].min()) / (
        df['CGPA'].max() - df['CGPA'].min() + 1e-8
    )

    return df

data_sent = normalize_features(data_sent)
```

```python
def get_recommendations(student_id, df=data_sent, embeddings=embeddings, top_n=10):

    df = df.reset_index(drop=True)

    target_vec = embeddings[student_id]

    # similarity
    sim_scores = cosine_similarity([target_vec], embeddings).flatten()

    # scoring
    final_scores = (
        0.80 * sim_scores +
        0.12 * df['proj_norm'].values +
        0.08 * df['cgpa_norm'].values
    )

    top_indices = np.argsort(final_scores)[::-1][:top_n]

    results = df.iloc[top_indices].copy()
    results['score'] = final_scores[top_indices]

    return results[['Name','Interested_domain','Skills','Projects','CGPA','score']]
```

```python
inertia = []
k_values = range(2, 15)

for k in k_values:
    model = KMeans(n_clusters=k, random_state=42)
    model.fit(embeddings)   # using sentence embeddings,         not tfidf
    inertia.append(model.inertia_)

plt.figure(figsize=(8,5))
plt.plot(k_values, inertia, marker='o')
plt.title("Elbow Method (Embeddings)")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.xticks(k_values)
plt.grid()
plt.show()
```

```python
def get_students_recomend(skills_req , project_title, top_n,vectorizer = sent_transformer , df = data_sent , embeddings = embeddings):
    query_text = skills_req + " " + project_title
    df_project = pd.DataFrame([query_text], columns=['final_text'])

    # 2. Encode
    target_vec = vectorizer.encode(df_project['final_text'].tolist())

    # 3. Cosine similarity
    cosine_sim = cosine_similarity(target_vec, embeddings).flatten()

    final_scores = (
        0.80 * cosine_sim +
        0.12 * df['proj_norm'].values +
        0.08 * df['cgpa_norm'].values
    )

    top_n = min(top_n, len(df))

    top_indices = np.argsort(final_scores)[::-1][:top_n]

    results = df.iloc[top_indices].copy()
    results['score'] = final_scores[top_indices]

    return results[['Name', 'Interested_domain', 'Skills', 'Projects', 'CGPA', 'score']]
```
