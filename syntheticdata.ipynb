```python
import os
import glob
import pandas as pd

files = glob.glob("data/*.csv")

if len(files) == 0:
    raise Exception("❌ No CSV files found in /data folder")

dfs = []

for file in files:
    df = pd.read_csv(file)
    df.columns = df.columns.str.strip()

    rename_map = {
        "Num_Projects": "Number of Projects",
        "Num Projects": "Number of Projects",
        "Num_Projects ": "Number of Projects"
    }

    df.rename(columns=rename_map, inplace=True)
    dfs.append(df)


combined_df = pd.concat(dfs, ignore_index=True)

combined_df.to_csv("final_clean_dataset.csv", index=False)

print("Shape:", combined_df.shape)
```
